## 2. Setup Test Images

Using remote URLs for testing.

```python
# Sample image URLs for testing (using placeholder images)
cats = [
    "https://placecats.com/200/200",
    "https://placecats.com/201/201",
    "https://placecats.com/202/202",
    "https://placecats.com/203/203",
]

dogs = [
    "https://placedog.net/200/200",
    "https://placedog.net/201/201",
    "https://placedog.net/202/202",
]

nature = [
    "https://picsum.photos/seed/nature1/200/200",
    "https://picsum.photos/seed/nature2/200/200",
    "https://picsum.photos/seed/nature3/200/200",
    "https://picsum.photos/seed/nature4/200/200",
    "https://picsum.photos/seed/nature5/200/200",
]

print(f"Cats: {len(cats)} images")
print(f"Dogs: {len(dogs)} images")
print(f"Nature: {len(nature)} images")
```

## 3. Test: Dictionary of Image Lists

Pass a dictionary where keys become tab names and values are lists of images.

```python
# Dictionary input - creates tabs with keys as tab names
images_dict = {
    "Cats": cats,
    "Dogs": dogs,
    "Nature": nature
}

print("Testing dict input with tabs...")
ipyplot.plot_images(images_dict, img_width=150)
```

```python
# Dict with max_images limiting images per tab
print("Testing dict input with max_images=2 per tab...")
ipyplot.plot_images(images_dict, img_width=150, max_images=2)
```

## 4. Test: List of Lists with Tabs Mode (Default)

Pass a list of lists where each inner list becomes a tab. Tab names are indices (0, 1, 2...) by default.

```python
# List of lists - default tabs mode with index as tab names
images_nested = [cats, dogs, nature]

print("Testing list of lists with tabs mode (default tab names: 0, 1, 2)...")
ipyplot.plot_images(images_nested, img_width=150)
```

```python
# List of lists with custom tab names via labels parameter
print("Testing list of lists with custom tab names via labels...")
ipyplot.plot_images(
    images_nested, 
    labels=["Cats", "Dogs", "Nature"],
    img_width=150
)
```

```python
# List of lists with max_images per tab
print("Testing list of lists tabs with max_images=2 per tab...")
ipyplot.plot_images(
    images_nested,
    labels=["Cats", "Dogs", "Nature"],
    img_width=150,
    max_images=2
)
```

## 5. Test: List of Lists with Grid Mode

Pass a list of lists where each inner list becomes a row. Empty cells are shown for shorter rows.

```python
# List of lists with grid mode - each inner list is a row
print("Testing list of lists with grid mode...")
ipyplot.plot_images(
    images_nested,
    img_width=150,
    nested_layout='grid'
)
```

```python
# Grid mode with row labels
print("Testing grid mode with row labels...")
ipyplot.plot_images(
    images_nested,
    labels=["Cats", "Dogs", "Nature"],
    img_width=150,
    nested_layout='grid'
)
```

```python
# Grid mode with uneven rows - demonstrates empty cell placeholders
uneven_images = [
    cats[:2],      # 2 images
    dogs,          # 3 images  
    nature,        # 5 images (longest row)
    cats[:1],      # 1 image
]

print("Testing grid mode with uneven rows (empty cell placeholders)...")
ipyplot.plot_images(
    uneven_images,
    labels=["Row 1 (2 imgs)", "Row 2 (3 imgs)", "Row 3 (5 imgs)", "Row 4 (1 img)"],
    img_width=120,
    nested_layout='grid'
)
```

## 6. Test: Standard Flat List (Backward Compatibility)

Ensure the original flat list behavior still works.

```python
# Standard flat list - should work as before
all_images = cats + dogs + nature

print("Testing standard flat list (backward compatibility)...")
ipyplot.plot_images(all_images, img_width=150, max_images=10)
```

```python
# Flat list with labels
flat_labels = ["cat"] * len(cats) + ["dog"] * len(dogs) + ["nature"] * len(nature)

print("Testing flat list with labels...")
ipyplot.plot_images(all_images, labels=flat_labels, img_width=150)
```

```python
# Dict with single key
single_key_dict = {
    "Only Cats": cats
}

print("Testing dict with single key...")
ipyplot.plot_images(single_key_dict, img_width=150)
```

## 9. Test: Tuple Inputs (image, label)

Images can be provided as tuples of `(image, label)`. When using this format:
- The label from the tuple is displayed instead of the default label
- The URL is NOT shown (only the custom label)

```python
# Flat list with tuple inputs - each image has its own label, URL not shown
images_with_labels = [
    (cats[0], "Fluffy Cat"),
    (cats[1], "Sleepy Cat"),
    (dogs[0], "Happy Dog"),
    (dogs[1], "Playful Dog"),
    (nature[0], "Mountain View"),
]

print("Testing flat list with tuple (image, label) inputs...")
print("Note: URLs should NOT be shown, only the custom labels")
ipyplot.plot_images(images_with_labels, img_width=150)
```

```python
# Mixed inputs: some tuples, some plain images
# Tuples will have custom labels (no URL), plain images show URL
mixed_images = [
    (cats[0], "Custom Label Cat"),  # Tuple: shows "Custom Label Cat", no URL
    cats[1],                         # Plain: shows URL
    (dogs[0], "Custom Label Dog"),  # Tuple: shows "Custom Label Dog", no URL
    dogs[1],                         # Plain: shows URL
]

print("Testing mixed inputs (tuples and plain images)...")
print("Tuples should show custom labels without URL, plain images show URL")
ipyplot.plot_images(mixed_images, img_width=150)
```

```python
# Grid mode with tuple inputs
grid_with_tuples = [
    [(cats[0], "Cat A"), (cats[1], "Cat B"), (cats[2], "Cat C")],
    [(dogs[0], "Dog A"), (dogs[1], "Dog B")],  # Shorter row - will have empty cell
    [(nature[0], "Nature 1"), (nature[1], "Nature 2"), (nature[2], "Nature 3")],
]

print("Testing grid mode with tuple (image, label) inputs...")
ipyplot.plot_images(
    grid_with_tuples,
    labels=["Cats Row", "Dogs Row", "Nature Row"],
    img_width=150,
    nested_layout='grid'
)
```

```python
# Dict with tuple inputs
dict_with_tuples = {
    "Cats": [(cats[0], "Whiskers"), (cats[1], "Mittens")],
    "Dogs": [(dogs[0], "Buddy"), (dogs[1], "Max"), (dogs[2], "Charlie")],
}

print("Testing dict input with tuple (image, label) values...")
ipyplot.plot_images(dict_with_tuples, img_width=150)
```
